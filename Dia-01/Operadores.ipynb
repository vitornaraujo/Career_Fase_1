```python
nome = input("Nome:")
dt_nacimento = input("Data de Nascimento:")
peso_atual  = input("Peso Atual:")
altura_em_metros = input("Altura em Metros:")
```

```python
imc = float(peso_atual) / float(altura_em_metros)**2
peso_ideal = lambda peso, altura: 18.5 <= (peso / altura**2) <= 24.9
 
```

```python
print(
f"Olá, {nome}! \n\
Você tem {dt_nacimento} anos. \n\
Seu IMC é {imc:.2f}. \n\
Você está dentro do peso ideal? {peso_ideal(float(peso_atual), float(altura_em_metros))}")
```

```python
import sqlite3


conn = sqlite3.connect("loja.db")
cursor = conn.cursor()
```

```python
import random
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS clientes (
    id INTEGER PRIMARY KEY,
    nome TEXT,
    idade INTEGER,
    telefone TEXT
)
""")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS endereco (
    id INTEGER PRIMARY KEY,
    id_cliente INTEGER,
    rua TEXT,
    numero TEXT,
    cidade TEXT,
    estado TEXT,
    cep TEXT,
    FOREIGN KEY (id_cliente) REFERENCES clientes(id)
)
""")
```

```python
nomes = [
    "Ana", "Bruno", "Carlos", "Daniela", "Eduardo",
    "Fernanda", "Gabriel", "Helena", "Igor", "Juliana"
]

ruas = [
    "Rua das Flores", "Av. Brasil", "Rua Central",
    "Rua das Palmeiras", "Av. Paulista"
]

cidades = ["São Paulo", "Rio de Janeiro", "Belo Horizonte", "Curitiba"]
estados = ["SP", "RJ", "MG", "PR"]

for _ in range(100):
    nome = random.choice(nomes)
    idade = random.randint(18, 80)
    telefone = f"(11)9{random.randint(1000,9999)}-{random.randint(1000,9999)}"

    cursor.execute(
        "INSERT INTO clientes (nome, idade, telefone) VALUES (?, ?, ?)",
        (nome, idade, telefone)
    )

    id_cliente = cursor.lastrowid

    rua = random.choice(ruas)
    numero = str(random.randint(1, 9999))
    cidade = random.choice(cidades)
    estado = random.choice(estados)
    cep = f"{random.randint(10000,99999)}-{random.randint(100,999)}"

    cursor.execute(
        """
        INSERT INTO endereco (id_cliente, rua, numero, cidade, estado, cep)
        VALUES (?, ?, ?, ?, ?, ?)
        """,
        (id_cliente, rua, numero, cidade, estado, cep)
    )
```

```python
cursor.execute("SELECT * FROM clientes LIMIT 5")
for row in cursor.fetchall():
    print(row)

   
```
